```python
import great_expectations as gx
from pathlib import Path
import pandas as pd
from great_expectations.core.validation_definition import ValidationDefinition

df = pd.read_csv("dataset1.csv")
df.drop(columns=["Unnamed: 0"], inplace=True)
UNIQUE_TRACK_GENRES = list(df["track_genre"].unique())

context = gx.get_context(mode="file", project_root_dir=Path("./new_context_folder"))
```

```python
source_folder = Path(".")
data_source_name = "my_filesystem_data_source"

data_source = context.data_sources.add_or_update_pandas_filesystem(
    name=data_source_name,
    base_directory=source_folder
)

asset_name = "music_csv_files"
try:
    file_csv_asset = data_source.get_asset(asset_name)
except LookupError:
    file_csv_asset = data_source.add_csv_asset(name=asset_name)

batch_definition_name = "dataset1.csv"
batch_definition_path = "dataset1.csv"

batch_definition = file_csv_asset.add_batch_definition_path(
    name=batch_definition_name,
    path=batch_definition_path
)

batch = batch_definition.get_batch()
print(batch.head())
```

```python
suite_name = "my_expectation_suite"

suite = context.suites.add_or_update(gx.ExpectationSuite(name=suite_name))

suite.save()

columns_no_nulls = ['track_id', 'artists', 'album_name',
 'track_name', 'popularity', 'duration_ms',
 'explicit', 'danceability', 'energy',
 'key', 'loudness', 'mode',
 'speechiness', 'acousticness', 'instrumentalness',
 'liveness', 'valence', 'tempo',
 'time_signature', 'track_genre']

for col in columns_no_nulls:
    suite.add_expectation(
        gx.expectations.ExpectColumnValuesToNotBeNull(column=col, mostly=1.0)
    )

suite.add_expectation(
    gx.expectations.ExpectColumnValueLengthsToBeBetween(
        column="track_id", 
        min_value=22, 
        max_value=22,
        meta={"notes": "Expecting track_id to have exactly 22 characters"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValueLengthsToBeBetween(
        column="artists",
        min_value=2,
        max_value=512,
        meta={"notes": "Expecting artists to have between 2 and 512 characters"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValueLengthsToBeBetween(
        column="album_name",
        min_value=2,
        max_value=512,
        meta={"notes": "Expecting album_name to have between 2 and 512 characters"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValueLengthsToBeBetween(
        column="track_name",
        min_value=2,
        max_value=512,
        meta={"notes": "Expecting track_name to have between 2 and 512 characters"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="popularity",
        min_value=0,
        max_value=100,
        meta={"notes": "Expecting popularity to be between 0 and 100"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="duration_ms",
        min_value=0,
        max_value=5237760,
        meta={"notes": "Expecting duration_ms to be between 0 and 5237760"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeOfType(
        column="explicit",
        meta={"notes": "Expecting explicit to be of type bool"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="danceability",
        min_value=0.0,
        max_value=1.0,
        meta={"notes": "Expecting danceability to be between 0.0 and 1.0"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="energy",
        min_value=0.0,
        max_value=1.0,
        meta={"notes": "Expecting energy to be between 0.0 and 1.0"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="key",
        min_value=0,
        max_value=11,
        meta={"notes": "Expecting key to be between 0 and 11"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="loudness",
        min_value=-45.0,
        max_value=5.0,
        meta={"notes": "Expecting loudness to be between -45.0 and 5.0"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="mode",
        min_value=0.0,
        max_value=1.0,
        meta={"notes": "Expecting mode to be between 0.0 and 1.0"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="speechiness",
        min_value=0.0,
        max_value=1.0,
        meta={"notes": "Expecting speechiness to be between 0.0 and 1.0"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="acousticness",
        min_value=0.0,
        max_value=1.0,
        meta={"notes": "Expecting acousticness to be between 0.0 and 1.0"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="instrumentalness",
        min_value=0.0,
        max_value=1.0,
        meta={"notes": "Expecting instrumentalness to be between 0.0 and 1.0"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="liveness",
        min_value=0.0,
        max_value=1.0,
        meta={"notes": "Expecting liveness to be between 0.0 and 1.0"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="valence",
        min_value=0.0,
        max_value=1.0,
        meta={"notes": "Expecting valence to be between 0.0 and 1.0"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="tempo",
        min_value=0.0,
        max_value=256.0,
        meta={"notes": "Expecting tempo to be between 0.0 and 256.0"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="time_signature",
        min_value=0,
        max_value=5,
        meta={"notes": "Expecting time_signature to be between 0 and 5"}
    )
)

suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeInSet(
        column="track_genre",
        value_set=UNIQUE_TRACK_GENRES,
        meta={"notes": "Expecting track_genre to be one of the 114 known genres"}
    )
)
```

```python
batch_definition = (
    context.data_sources.get(data_source_name)
    .get_asset(asset_name)
    .get_batch_definition(batch_definition_name)
)

validation_name = "validation_run_1"

validation_definition = ValidationDefinition(
    name=validation_name,
    data=batch_definition,
    suite=suite
)
```
